```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

def kaprekar_steps(n):
    """Повертає список кроків процесу Капрекара для числа n (4 цифри)."""
    steps = [n]
    while n != 6174:
        digits = sorted(f"{n:04d}")
        asc = int("".join(digits))
        desc = int("".join(reversed(digits)))
        n = desc - asc
        steps.append(n)

    return steps


def visualize_kaprekar(start=1234):
    """Візуалізує процес досягнення сталої Капрекара (6174) для заданого числа."""
    steps = kaprekar_steps(start)

    fig, ax = plt.subplots(figsize=(8, 5))
    fig.suptitle(f"Стала Капрекара (6174) — початкове число: {start}", fontsize=14, fontweight="bold")

    ax.plot(steps, marker="o", color="steelblue", linewidth=2, markersize=8)
    ax.axhline(y=6174, color="crimson", linestyle="--", linewidth=1.5, label="6174 (стала)")
    for i, v in enumerate(steps):
        ax.annotate(str(v), (i, v), textcoords="offset points", xytext=(0, 10), ha="center", fontsize=9)
    ax.set_title("Послідовність кроків")
    ax.set_xlabel("Крок")
    ax.set_ylabel("Значення")
    ax.set_xticks(range(len(steps)))
    ax.legend()
    ax.grid(True, linestyle="--", alpha=0.5)

    plt.tight_layout()
    plt.show()
    print(f"Число {start} → стала Капрекара за {len(steps)-1} крок(ів): {' → '.join(map(str, steps))}")
```

```python
visualize_kaprekar(1234)
```
